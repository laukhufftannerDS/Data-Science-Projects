Apply Broyden I with starting guesses $x_0$ = (1,1) and $A_0$ = I to the systems in exercise 3. Report the solutions with as much accuracy as possible and the number of steps required.

a.
\begin{cases}
u^2 + v^2 = 1 \\
(u - 1)^2 + v^2 = 1
\end{cases}
b.
\begin{cases}
u^2 + 4v^2 = 4 \\
4u^2 + v^2 = 4
\end{cases}
c.
\begin{cases}
u^2 - 4v^2 = 4 \\
(u - 1)^2 + v^2 = 4
\end{cases}

```python
import numpy as np
def broyden(F,x0,A0=None,tol=1e-12,max_iter=100):
    x = np.array(x0,dtype=float)
    n = len(x)
    if A0 is None:
        A = np.eye(n)
    else:
        A = np.array(A0,dtype=float)
    for k in range(max_iter):
        Fx = np.array(F(x))
        if np.linalg.norm(Fx,ord=2) < tol:
            return x,k
        s = np.linalg.solve(A,-Fx)
        x_new = x+s
        Fx_new = np.array(F(x_new))
        y = Fx_new-Fx
        A = A+np.outer((y-A@s),s)/(s@s)
        x = x_new
    raise RuntimeError("Broyden did not converge within max iterations")
def F_a(X):
    u,v = X
    return [u**2+v**2-1,(u-1)**2+v**2-1]
def F_b(X):
    u,v = X
    return [u**2+4*v**2-4,4*u**2+v**2-4]
def F_c(X):
    u,v = X
    return [u**2-4*v**2-4,(u-1)**2+v**2-4]
systems = {'a':F_a,'b':F_b,'c':F_c}
x0 = [1,1]
for name, F in systems.items():
    sol, steps = broyden(F,x0)
    print(f"System {name}: solution = {sol}, steps = {steps}")
```

Apply Broyden II with starting guesses $x_0$ = (1,1) and $B_0$ = I to the systems in exercise 3. Report the solutions with as much accuracy as possible and the number of steps required.

a.
\begin{cases}
u^2 + v^2 = 1 \\
(u - 1)^2 + v^2 = 1
\end{cases}
b.
\begin{cases}
u^2 + 4v^2 = 4 \\
4u^2 + v^2 = 4
\end{cases}
c.
\begin{cases}
u^2 - 4v^2 = 4 \\
(u - 1)^2 + v^2 = 4
\end{cases}

```python
import numpy as np
def broyden_ii(F,x0,B0=None,tol=1e-12,max_iter=1000):
    x = np.array(x0,dtype=float)
    n = len(x)
    if B0 is None:
        B = np.eye(n)
    else:
        B = np.array(B0,dtype=float)
    for k in range(max_iter):
        Fx = np.array(F(x))
        if np.linalg.norm(Fx,ord=2) < tol:
            return x,k
        s = -B@Fx
        x_new = x+s
        Fx_new = np.array(F(x_new))
        y = Fx_new-Fx
        B = B+np.outer((s-B@y),s)/(s@y)
        x = x_new
    raise RuntimeError("Broyden II did not converge within max iterations")
def F_a(X):
    u,v = X
    return [u**2+v**2-1,(u-1)**2+v**2-1]
def F_b(X):
    u,v = X
    return [u**2+4*v**2-4,4*u**2+v**2-4]
def F_c(X):
    u,v = X
    return [u**2-4*v**2-4,(u-1)**2+v**2-4]
systems = {'a':F_a,'b':F_b,'c':F_c}
x0 = [1,1]
B0 = np.eye(2)
for name, F in systems.items():
    try:
        sol, steps = broyden_ii(F,x0,B0)
        print(f"System {name}: solution = {sol}, steps = {steps}")
    except RuntimeError:
        print(f"System {name}: Broyden II did not converge")
```
