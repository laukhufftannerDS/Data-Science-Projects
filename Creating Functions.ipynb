# Differentiate Function
This cell implements the numerical derivative. 

This function computes a numerical approximation of the derivative using a symmetric difference quotient.  Use the central difference formula as h approaches 0 to get a very accurate approximation and then use try and except blocks to print errors.
$$
f'(x) = \lim_{h \to 0} \frac{f(x+h) - f(x-h)}{2h}
$$

```python
def differentiate(f,x):
    h=1e-6
    if not callable(f):
        print("Error: f must be a function that can be called like f(x)")
        return None
    try:
        x=float(x)
    except Exception:
        print("Error: x must be a number")
        return None
    try:
        fxh=f(x+h)
        fxh2=f(x-h)
    except Exception as e:
        print("Error while evaluating f(x):",e)
        return None
    try:
        value = (fxh-fxh2)/(2*h)
        return round(value,5)
    except Exception as e:
        print("Error while computing the derivative:",e)
        return None

def f1(x): return x*x
print(differentiate(f1,2))      # expect about 4

def f2(x): return x**3
print(differentiate(f2,1))      # expect about 3

def f3(x): return 5*x**4 - 3*x
print(differentiate(f3,0))      # expect about -3

def f4(x): return x**5
print(differentiate(f4,-1))     # expect about 5

def f5(x): return 7
print(differentiate(f5,10))     # constant → derivative 0

print(differentiate("abc",1))   # error: function is not callable
print(differentiate(f1,"xyz"))  # error: x is not numeric

def f8(x): return x/0
print(differentiate(f8,2))      # error: division by zero inside f
```

2. integrate(f,a,b) which returns a numerical approximation of the integral of a function
f between a and b using Simpson’s rule with 5 significant digits of accuracy.

```python
def integrate(f,a,b):
    try:
        if not callable(f):
            raise TypeError("f must be a callable function")
        if not isinstance(a,(int,float)):
            raise TypeError("a must be a number")
        if not isinstance(b,(int,float)):
            raise TypeError("b must be a number")
        if a == b:
            return 0.0
        if b < a:
            a,b = b,a
        n = 2
        max_iterations = 1000
        tol = 0.5*10**-5
        previous = None
        while True:
            h = (b-a)/n
            x0 = f(a)+f(b)
            odd_sum = 0
            for i in range(1,n,2):
                odd_sum += f(a+i*h)
            even_sum = 0
            for i in range(2,n,2):
                even_sum += f(a+i*h)
            integral = (h/3)*(x0+4*odd_sum+2*even_sum)
            if previous is not None and abs(integral-previous) < tol:
                return round(integral,5)
            previous = integral
            n *= 2
            max_iterations -= 1
            if max_iterations <= 0:
                print("Maximum iterations reached")
                return round(integral,5)
    except Exception as e:
        print(f"{e}")
        return None
def f1(x):
    return x
def f2(x):
    return x**2
def f3(x):
    import math
    return math.sin(x)
def f4(x):
    import math
    return math.exp(x)
def f5(x):
    return x**3+2*x+1
print(integrate(f1,0,1))
print(integrate(f2,0,1))
print(integrate(f3,0,3.14159))
print(integrate(f4,0,1))
print(integrate(f5,2,2))
integrate(5,0,1)
integrate(f1,None,1)
integrate(f1,0,None)
```

This code finds the integral using Simpson's Rule to 5 significant digits. It starts by error checking to make sure the inputs are integers or floats and makes sure the function is usable. It also makes sure that a does not equal b because that will always equal 0. Then, it sets up Simpson's rule looking for 5 significant digits and creates the loop looking at the odd and even x-values. Finally it computes the approximation and checks for accuracy, rounding to the 5th decimal place. It also has a max iteration to prevent the function from infinite looping. 

# Newton's Method Function
This function approximates roots of equations using iterative updates and numerical derivatives.  Utilizes the differentiate function, with f being a starting guess, a = x, and 1000 iterations
$$
x_{n+1} = x_n - \frac{f(x_n)}{f'(x_n)}
$$

```python
def newtons(f,a,n=50,tol=1e-10):     
    h=1e-6
    # error checking
    if not callable(f):
        print("Error: f must be a function that can be called like f(x)")
        return None
    try:
        x=float(a)
    except Exception:
        print("Error: starting value a must be a number")
        return None
    try:
        n_int=int(n)
        if n_int<=0:
            print("Error: n must be a positive integer")
            return None
    except Exception:
        print("Error: n must be an integer")
        return None
    try:
        tol=float(tol)
        if tol<=0:
            print("Error: tol must be positive")
            return None
    except Exception:
        print("Error: tol must be a number")
        return None

    for i in range(n_int):
        try:
            fx=f(x)
        except Exception as e:
            print("Error while evaluating f(x):",e)
            return None
        try:
            dfx=(f(x+h)-f(x-h))/(2*h)
        except Exception as e:
            print("Error while evaluating derivative:",e)
            return None
        if dfx==0:
            print("Error: derivative became zero.")
            return None
        x_new=x-fx/dfx
        if abs(x_new-x)<tol:
            return round(x_new,5)
        x=x_new
    print("Warning: maximum iterations reached; result may not have converged")
    return round(x,5)

def g1(x): return x*x - 4
print(newtons(g1,1))        # expect about 2

def g2(x): return x**3 - 8
print(newtons(g2,2))        # expect about 2

def g3(x): return (x-3)**2
print(newtons(g3,3))        # flat derivative at x=3, derivative becomes 0, returns None

def g4(x): return x**4 - 16
print(newtons(g4,2))        # expect about 2

def g5(x): return x
print(newtons(g5,0))        # root is 0

print(newtons("notfunc",1))    # f is a string, error caught, returns None
print(newtons(g1,"xyz"))       # starting x is string, error caught, returns None

def g8(x): raise ValueError("boom")
print(newtons(g8,1))           # g8 raises Value Error, caught, returns None

def g9(x): return 1
print(newtons(g9,5))           # derivative about 0, prints derivative error, returns None

print(newtons(g1,10,n=3))      # only 3 steps, likely not converged
```

4. powerSeries(coefficients, a) takes in a list of numbers and returns a function which
represents the power series centered at a with those numbers as f (a), f ′(a), f ′′(a), . . ..
For example, passing in the list [1, 3, 5, 7] and 2 would cause the function to return
f (x) = 1 + 3
1!(x − 2) + 5
2!(x − 2)2 + 7
3!(x − 2)3

```python
def powerSeries(coefficients,a):
    try:
        if not isinstance(coefficients,list):
            raise TypeError("coefficients must be a list of numbers")
        if not all(isinstance(c,(int,float)) for c in coefficients):
            raise TypeError("All coefficients must be numbers")
        if not isinstance(a,(int,float)):
            raise TypeError("a must be a number")
        def factorial(n):
            if n < 0:
                raise ValueError("Factorial is not defined for negative numbers")
            result = 1
            for i in range(1,n+1):
                result *= i
            return result
        series_terms = []
        for i, coeff in enumerate(coefficients):
            if i == 0:
                term = f"{coeff}"
            elif i == 1:
                term = f"{coeff}/1!*(x-{a})"
            else:
                term = f"{coeff}/{i}!*(x-{a})^{i}"
            series_terms.append(term)
        series_string = "+".join(series_terms)
        print(f"f(x) = {series_string}")
        def series(x):
            if not isinstance(x,(int,float)):
                raise TypeError("x must be a number")
            total = 0
            for i, coeff in enumerate(coefficients):
                total += coeff/factorial(i)*(x-a)**i
            return total
        return series
    except Exception as e:
        print(f"{e}")
        return None
ps1 = powerSeries([1,3,5,7],2)
print("f(3) =", ps1(3))
ps2 = powerSeries([4],0)
print("f(10) =", ps2(10))
ps3 = powerSeries([2,5],1)
print("f(3) =", ps3(3))
ps4 = powerSeries([1,-2,3,-4,5],-1)
print("f(0) =", ps4(0))
ps5 = powerSeries([3,0,4,0,5],5)
print("f(6) =", ps5(6))
ps_err1 = powerSeries("list",0)
ps_err2 = powerSeries([1,"a",3],0)
ps_err3 = powerSeries([1,2,3],"a")
ps_err4 = powerSeries([1,2,3],0)
try:
    print(ps_err4("x"))
except Exception as e:
    print(f"{e}")
```

# Logarithm Function (Power Series Expansion)
Computes log base *b* of *x* using a custom natural log implementation with scaling.  Uses the Maclaurin 
$$
\ln\left(\frac{1+z}{1-z}\right)
= 2\left( z + \frac{z^3}{3} + \frac{z^5}{5} + \cdots \right)
$$
where
$$
z = \frac{y - 1}{y + 1}
$$
and change of base formula is 
$$
\log_b(x) = \frac{\ln(x)}{\ln(b)}
$$

```python
def log(x, b=2.7182818):
    # Error checking
    try:
        x = float(x)
        b = float(b)
    except:
        print('Error: x and base b must be numbers')
        return None

    if x <= 0:
        print('Error: log undefined for non-positive x')
        return None
    if b <= 0 or b == 1:
        print('Error: invalid base for logarithm')
        return None

    # Natural log helper using series with scaling
    def ln_sub(y):
        # scale y into (0.5, 2]
        count = 0
        while y > 2.0:
            y /= 2.0
            count += 1
        while y < 0.5:
            y *= 2.0
            count -= 1

        # series for ln(y)
        z = (y - 1.0) / (y + 1.0)
        term = z
        total = 0.0
        k = 1.0
        while abs(term) > 1e-7:
            total += term / k
            term *= z * z
            k += 2.0
        ln_y = 2.0 * total
        ln2 = 0.6931471805599453
        return ln_y + count * ln2
    ln_x = ln_sub(x)
    ln_b = ln_sub(b)
    value = ln_x / ln_b
    return round(value,5)

print(log(1))             # expect 0
print(log(10,10))         # expect 1
print(log(8,2))           # expect 3
print(log(1000))         # expect about 6.907755
print(log("20",10))       # string converted to valid
print(log(0,10))          # error: log undefined for non-positive x
print(log(-1,10))         # error: x must be positive
print(log(10,1))          # error: base cannot be 1
print(log(10,-2))         # error: base must be >0
print(log("abc",10))      # error: x not numeric
```

## Cosine Function
Computes cosine using argument reduction and a Taylor series expansion.  Use while loop for the domain.  Then use standard Maclaurin series.  And then it stops adding terms once it gets small enough.
$$
\cos(x)
= 1 - \frac{x^2}{2!}
+ \frac{x^4}{4!}
- \frac{x^6}{6!}
+ \cdots
$$

```python
def cos(x):
    try:
        x = float(x)
    except:
        print('Error: x must be a number')
        return None
    pi = 3.141592653589793
    two_pi = 2 * pi

    # reduce x into [-pi, pi]
    while x > pi:
        x -= two_pi
    while x < -pi:
        x += two_pi

    # Taylor series for cos(x)
    term = 1.0
    total = 1.0
    n = 0

    while True:
        n += 1
        term *= - x * x / ((2 * n - 1) * (2 * n))
        total += term
        if abs(term) < 1e-7:
            break
        if n > 100:
            break
    return round(total,5)

print(cos(0))             # expect about 1
print(cos(1))             # expect about 0.54030
print(cos(-1))            # expect about 0.54030
print(cos(0.5))           # expect about 0.87758
print(cos("1.0"))         # string converts to valid (.54)


print(cos("abc"))         # error: not numeric

print(cos(1000))            # large input, approximation (.56237)
print(cos(2))             # expect about -0.41615
print(cos(0.1))           # expect about 0.99500
print(cos(3))             # expect about -0.98999
```

## Secant Function
Computes sec(x) = 1 / cos(x) with undefined checks near cos(x)=0.  Takes the cosine function that we defined.  If cosx gets too small, secx becomes 
$$
\sec(x) = \frac{1}{\cos(x)}
$$

```python
def sec(x):
    # Compute cos(x) using our own cosine implementation
    c = cos(x)

    # If cos(x) failed, propagate the error
    if c is None:
        return None

    # If cos(x) is too close to 0, sec(x) is undefined
    if abs(c) < 1e-7:
        print('Error: sec(x) undefined because cos(x) is too close to 0')
        return None

    # Otherwise, return the reciprocal
    return round(1.0 / c,5)

pi=3.141592653589793238462643383279
print(sec(0))             # expect 1
print(sec(1))             # about 1.85082
print(sec(pi/2))           # Should produce error because cos(x) is too close to zero
print(sec(0.5))           # about 1.13949
print(sec("1.0"))         # string converts to valid


print(sec("abc"))         # error: not numeric

print(sec(10))            # about -1.19179
print(sec(2))             # about -2.40299
print(sec(0.1))           # about 1.00504
print(sec(3))             # about -1.01011
```

# Cotangent Function cot(x) = cos(x) / sin(x)
Computes cot(x) = cos(x)/sin(x) and checks for undefined values when sin(x) is near zero. 
$$
\cot(x) = \frac{\cos(x)}{\sin(x)}
$$

```python
def sin(x):
    try:
        if not isinstance(x,(int,float)):
            raise TypeError("x must be a number")
        pi = 3.141592653589793
        x = x%(2*pi)
        term = x
        sum_ = term
        n = 1
        tol = 0.5*10**-5
        while True:
            term *= -1*x**2/((2*n)*(2*n+1))
            sum_ += term
            if abs(term) < tol:
                break
            n += 1
            if n > 1000:
                print("Warning: Maximum iterations reached; result may be less accurate")
                break
        return round(sum_,5)
    except Exception as e:
        print(f"Error in sin function: {e}")
        return None
def cot(x):
    # Compute sin(x) and cos(x)
    s = sin(x)
    c = cos(x)

    # If either failed, propagate the error
    if s is None or c is None:
        return None

    # If sin(x) is too close to 0, cot(x) is undefined
    if abs(s) < 1e-7:
        print('Error: cot(x) undefined because sin(x) is too close to 0')
        return None

    # Otherwise, return cos(x) / sin(x)
    return round(c / s,5)
print(cot(1))             # expect about 0.64209
print(cot(0.5))           # expect about 1.83049
print(cot(-1))            # expect about -0.64209
print(cot(2))             # expect about -0.45766
print(cot("1.0"))         # string converts to valid
print(cot(pi))            # Undefined: sin(x) to close to 0

print(cot("abc"))         # error: not numeric

print(cot(10))            # expect about 1.54235
print(cot(3))             # expect large negative (-7.0152)
print(cot(0.1))           # expect large positive (9.9669)
print(cot(4))             # expect about 0.86369
```

# Inverse Cosine Function acos(x)
Computes arccos(x) using the bisection method on cos(y) = x. Use our arcsin function and the trig identity formula.  Series for arcsin converges for absolute value of x is less than or equal to 1 and stop when terms are smaller than 10^-7
$$
\arccos(x) = \frac{\pi}{2} - \arcsin(x)
$$

```python
def acos(x):
    try:
        x = float(x)
    except:
        print('Error: x must be a number')
        return None

    # Domain check
    if x < -1 or x > 1:
        print('Error: acos(x) undefined for |x| > 1')
        return None

    # Compute asin(x) using Taylor series
    def asin_series(x):
        try:
            if abs(x) > 1:
                raise ValueError("asin(x) is undefined for |x| > 1")

            def factorial(n):
                if n < 0:
                    raise ValueError("Factorial not defined for negative numbers")
                result = 1
                for i in range(1,n+1):
                    result *= i
                return result

            term = x
            sum_ = term
            n = 1
            tol = 0.5*10**-5
            while True:
                coeff = factorial(2*n) / ((4**n)*(factorial(n)**2)*(2*n+1))
                term = coeff * (x**(2*n+1))
                sum_ += term
                if abs(term) < tol:
                    break
                n += 1
                if n > 1000:
                    print("Warning: Maximum iterations reached; result may be less accurate")
                    break
            return round(sum_,5)
        except Exception as e:
            print(f"Error in asin function: {e}")
            return None

    pi = 3.141592653589793
    asin_x = asin_series(x)
    if asin_x is None:
        return None
    value = pi/2 - asin_x
    return round(value,5)
print(acos(1))            # expect 0
print(acos(0))            # expect ~1.57080
print(acos(0.5))          # expect ~1.04720
print(acos(-0.5))         # expect ~2.09440

print(acos(2))            # error: |x|>1 not allowed
print(acos(-2))           # error: |x|>1 not allowed

print(acos("0.5"))        # string converts to valid
print(acos("abc"))        # error: not numeric

print(acos(1e-8))         # expect ~1.57080
```

16. atan(x) returns the arctangent of x in radians with 5 significant digits of accuracy.

```python
def atan(x):
    try:
        if not isinstance(x,(int,float)):
            raise TypeError("x must be a number")
        pi = 3.141592653589793
        tol = 0.5*10**-5
        if abs(x) <= 1:
            term = x
            sum_ = term
            n = 1
            while True:
                term = (-1)**n*x**(2*n+1)/(2*n+1)
                sum_ += term
                if abs(term) < tol:
                    break
                n += 1
                if n > 100000:
                    print("Maximum iterations reached")
                    break
            return round(sum_,5)
        else:
            sign = 1 if x > 0 else -1
            return round(sign*pi/2-atan(1/x),5)
    except Exception as e:
        print(f"{e}")
        return None
pi = 3.141592653589793
print(atan(0))
print(atan(1))
print(atan(-1))
print(atan(0.5))
print(atan(5))
print(atan(-10))
atan("hello")
atan(None)
atan([1,2,3])
atan(1e10)
```
